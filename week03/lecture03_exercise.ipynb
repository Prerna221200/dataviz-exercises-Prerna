```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)
df = pd.read_csv('../data/co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

## Task — Multi-Series Line Chart with Highlight

**What to build:** A line chart showing CO2 emissions over time for **all Asian countries** in the dataset, with one country highlighted.

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country and its story

```python
# Task — Multi-series line with highlight
# YOUR CODE HERE

import plotly.express as px
import plotly.graph_objects as go

# Filter the dataset to include only Asian countries
asia_df = df[df["Region"] == "Asia"].copy()

# Choose the country to highlight
highlight_country = "China"

# Create a colour map:
# Highlight China in red and all other countries in light grey
color_map = {
    country: "darkblue" if country == highlight_country else "#E5E5E5"
    for country in asia_df["Country"].unique()
}

# Create the multi-series line chart
fig = px.line(
    asia_df,
    x="Year",
    y="CO2_Mt",
    color="Country",
    color_discrete_map=color_map,
    labels={
        "Year": "Year",
        "CO2_Mt": "CO₂ emissions (million tonnes)"
    },
    title="China Dominates Asia's Recorded CO₂ Emissions",
    template="plotly_white",
    height=500
)

# Make all country lines thin and remove the legend
fig.update_traces(
    line=dict(width=1.2),
    showlegend=False
)

# Make the highlighted country's line thicker
fig.update_traces(
    line=dict(width=3),
    selector=dict(name=highlight_country)
)

# Find the final data point for China
highlight_data = (
    asia_df[asia_df["Country"] == highlight_country]
    .sort_values("Year")
)

# Add a direct label at the end of China's line
if not highlight_data.empty:
    last_point = highlight_data.iloc[-1]

    fig.add_annotation(
        x=last_point["Year"],
        y=last_point["CO2_Mt"],
        text=highlight_country,
        showarrow=False,
        xanchor="left",
        xshift=8,
        font=dict(
            color="darkblue",
            size=12,
            family="Arial"
        )
    )
else:
    print(f"Warning: {highlight_country} was not found in the Asian dataset.")

# Clean up the chart layout
fig.update_layout(
    title=dict(
        text="China Dominates Asia's Recorded CO₂ Emissions",
        x=0,
        xanchor="left",
        font=dict(size=20)
    ),
    xaxis=dict(
        showgrid=False,
        title="Year"
    ),
    yaxis=dict(
    title="CO₂ emissions (million tonnes)",
    gridcolor="#EEEEEE",
    zeroline=False,
    tickformat=",.0f"
),
    plot_bgcolor="white",
    paper_bgcolor="white",
    margin=dict(l=70, r=110, t=90, b=60),
    hovermode="x unified"
)

# Display the chart
fig.show()
```
